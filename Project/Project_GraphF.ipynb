```python
import numpy as np
import matplotlib.pyplot as plt

#Filename, als enige editen
fileName = "Draaiing2.csv"

#Load header
header = np.loadtxt(fileName, delimiter=',', max_rows=1, dtype=str)
print(header)

#Load rest data
data = np.loadtxt(fileName, delimiter=',', skiprows=1)
data = data.T

#2d to data
dataTime = data[0]  
dataAccX = data[1]  
dataAccY = data[2]  
dataAccZ = data[3]

#Plot x
plt.figure(figsize=(10, 6))

plt.plot(dataTime[1000:3482], dataAccX[1000:3482], label='x')

plt.xlabel('Time (s)')
plt.ylabel(' Acceleration (m/s^2)')
plt.title("x-as")
plt.legend()
plt.grid(True)
plt.ylim(-6, 2)
plt.show()

#Plot y
plt.figure(figsize=(10, 6))

plt.plot(dataTime[1000:3482], dataAccY[1000:3482], label='y', c='green')

plt.xlabel('Time (s)')
plt.ylabel(' Acceleration (m/s^2)')
plt.title("y-as")
plt.legend()
plt.grid(True)
plt.ylim(-6, 2)
plt.show()

#Plot z
plt.figure(figsize=(10, 6))

plt.plot(dataTime[1000:3482], dataAccZ[1000:3482], label='Z', c='m')

plt.xlabel('Time (s)')
plt.ylabel(' Acceleration (m/s^2)')
plt.title("Z-as")
plt.legend()
plt.grid(True)
plt.ylim(-6, 2)
plt.show()
```
